```python
from datetime import date

# Aqui em cima você coloca/importa suas classes:
# from modelo import Sistema, Projeto, Tarefa, Usuario

def ler_data():
    """Lê uma data no formato dd/mm/aaaa e devolve um objeto date."""
    texto = input("Data (dd/mm/aaaa): ")
    dia, mes, ano = map(int, texto.split("/"))
    return date(ano, mes, dia)

def menu_nao_logado():
    print("\n=== MENU (não logado) ===")
    print("1 - Registrar usuário")
    print("2 - Fazer login")
    print("0 - Sair")

def menu_logado(usuario):
    print(f"\n=== MENU (logado como {usuario.nome}) ===")
    print("1 - Criar projeto")
    print("2 - Listar projetos")
    print("3 - Criar tarefa em um projeto")
    print("4 - Listar tarefas de um projeto")
    print("5 - Listar minhas tarefas")
    print("6 - Sair da conta")
    print("0 - Encerrar programa")

def escolher_projeto(sistema):
    nome = input("Nome do projeto: ")
    try:
        projeto = sistema.buscar_projeto(nome)
        return projeto
    except LookupError as e:
        print("Erro:", e)
        return None

def main():
    sistema = Sistema()
    usuario_logado = None

    while True:
        if usuario_logado is None:
            # MENU PARA QUEM NÃO ESTÁ LOGADO
            menu_nao_logado()
            opcao = input("Escolha uma opção: ")

            if opcao == "1":
                print("\n--- Registrar usuário ---")
                nome = input("Nome: ")
                email = input("E-mail: ")
                senha = input("Senha: ")
                try:
                    usuario = sistema.registrar_usuario(nome, email, senha)
                    print("Usuário registrado com sucesso:", usuario.nome)
                except ValueError as e:
                    print("Erro:", e)

            elif opcao == "2":
                print("\n--- Login ---")
                email = input("E-mail: ")
                senha = input("Senha: ")
                try:
                    usuario_logado = sistema.autenticar_usuario(email, senha)
                    print("Login realizado com sucesso!")
                except ValueError as e:
                    print("Erro:", e)

            elif opcao == "0":
                print("Encerrando o sistema...")
                break

            else:
                print("Opção inválida. Tente novamente.")

        else:
            # MENU PARA QUEM ESTÁ LOGADO
            menu_logado(usuario_logado)
            opcao = input("Escolha uma opção: ")

            if opcao == "1":
                print("\n--- Criar projeto ---")
                nome = input("Nome do projeto: ")
                descricao = input("Descrição: ")
                try:
                    projeto = sistema.criar_projeto(nome, descricao)
                    print("Projeto criado com sucesso:", projeto.nome)
                except ValueError as e:
                    print("Erro:", e)

            elif opcao == "2":
                print("\n--- Listar projetos ---")
                if not sistema.projetos:
                    print("Nenhum projeto cadastrado.")
                else:
                    for p in sistema.projetos:
                        print(f"- {p.nome} ({p.descricao})")

            elif opcao == "3":
                print("\n--- Criar tarefa em um projeto ---")
                projeto = escolher_projeto(sistema)
                if projeto is None:
                    continue
                titulo = input("Título da tarefa: ")
                descricao = input("Descrição: ")
                print("Data de vencimento:")
                try:
                    vencimento = ler_data()
                    # aqui você poderia validar título/descrição/vencimento e dar raise se quiser
                    tarefa = sistema.criar_tarefa(projeto, titulo, descricao, vencimento)
                    # atribuindo a tarefa ao usuário logado:
                    tarefa.atribuir(usuario_logado)
                    print("Tarefa criada e atribuída a", usuario_logado.nome)
                except ValueError as e:
                    print("Erro:", e)

            elif opcao == "4":
                print("\n--- Listar tarefas de um projeto ---")
                projeto = escolher_projeto(sistema)
                if projeto is None:
                    continue
                tarefas = projeto.listar_tarefas()
                if not tarefas:
                    print("Nenhuma tarefa neste projeto.")
                else:
                    for t in tarefas:
                        print(f"- {t.titulo} ({t.status}) - vence em {t.vencimento}")

            elif opcao == "5":
                print("\n--- Minhas tarefas ---")
                tarefas = sistema.listar_tarefas_do_usuario(usuario_logado)
                if not tarefas:
                    print("Você não tem tarefas atribuídas.")
                else:
                    for t in tarefas:
                        print(f"- {t.titulo} ({t.status})")

            elif opcao == "6":
                print("Saindo da conta...")
                usuario_logado = None

            elif opcao == "0":
                print("Encerrando o sistema...")
                break

            else:
                print("Opção inválida. Tente novamente.")

if __name__ == "__main__":
    main()

```
